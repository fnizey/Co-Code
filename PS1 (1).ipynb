### Exercise 14. Iteration and convergence

A firm's capital stock depreciates at a rate $\delta$ each year, and the firm invests a fixed
amount $i$ each year. Capital therefore evolves according to

$$k_{t+1} = (1-\delta)\,k_t + i, \qquad k_0 = 0, \quad \delta = 0.1, \quad i = 0.3 .$$

A **steady state** (or **fixed point**) $k^{*}$ is a capital stock that reproduces itself,

$$k^{*} = (1-\delta)k^{*} + i \qquad\Longleftrightarrow\qquad k^{*} = \frac{i}{\delta}.$$

Subtracting this from the equation above gives the *deviation* from the steady state:

$$k_{t+1} - k^{*} = (1-\delta)\,(k_t - k^{*})
\qquad\Longrightarrow\qquad
k_t - k^{*} = (1-\delta)^{t}\,(k_0 - k^{*}).$$

So the distance to $k^{*}$ is multiplied by $(1-\delta)$ every single period, and the iteration
converges, from **any** starting point, exactly when $|1-\delta| < 1$. That inequality is the
whole idea behind every iterative method in this course.

### Exercise 15. Fixed points and contraction

The equation

$$x^{2} = x + 2$$

has the two solutions $x=2$ and $x=-1$. It can be rearranged into a fixed-point problem
$x = g(x)$ in more than one way. Two natural choices are

$$x_{n+1} = \sqrt{x_n + 2}
\qquad\text{and}\qquad
x_{n+1} = x_n^{2} - 2 .$$

Both leave $x^{*}=2$ unchanged, so both have it as a fixed point. Whether an iteration of this kind
actually *finds* it depends on the slope of the map at that point. If

$$\bigl|g'(x^{*})\bigr| < 1$$

then the distance to $x^{*}$ shrinks at every step, the map is a **contraction**, and close to
the fixed point the errors satisfy

$$\frac{\bigl|x_{n+1}-x^{*}\bigr|}{\bigl|x_{n}-x^{*}\bigr|} \;\longrightarrow\; \bigl|g'(x^{*})\bigr| .$$

For the two rules above,

$$g(x) = \sqrt{x+2}, \qquad g'(x) = \frac{1}{2\sqrt{x+2}}, \qquad g'(2) = \tfrac14,$$

$$h(x) = x^{2}-2, \qquad h'(x) = 2x, \qquad h'(2) = 4 .$$

(These two maps are inverses of one another, $h(g(x)) = x$, which is why their slopes at $x^{*}$
are reciprocals. You met both of them, with pen and paper, in Question 4 of PS0.)
